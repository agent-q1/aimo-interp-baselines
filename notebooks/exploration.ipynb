# Exploring the robustness datasets

Load both CSV files into pandas DataFrames. `df_agg` contains the aggregated records; `df_final` contains the category-specific and OVERALL records used by the baseline.

```python
from pathlib import Path

import pandas as pd

# Support launching Jupyter from either the repo root or notebooks/.
repo_root = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / 'data' / 'math-robust-final.csv').is_file()
)
df_agg = pd.read_csv(repo_root / 'data' / 'math-robust-agg.csv')
df_final = pd.read_csv(repo_root / 'data' / 'math-robust-final.csv')
```

```python
pd.DataFrame({
    'file': ['math-robust-agg.csv', 'math-robust-final.csv'],
    'rows': [len(df_agg), len(df_final)],
    'columns': [df_agg.shape[1], df_final.shape[1]],
    'unique_problems': [df_agg['problem_id'].nunique(), df_final['problem_id'].nunique()],
})
```

## Aggregated records

```python
# Count number of comma in each permutation type and add 1 to get the number of permutations
df_agg['permutation_type'].apply(lambda x: x.count(',') + 1 if isinstance(x, str) else 1).sum()
```

```python
# Extract the number of permutations if it's a list. If it's a string let it just be 1
df_agg['permutation_type'].apply(lambda x: len(eval(x)) if isinstance(x, list) else 1)
```
